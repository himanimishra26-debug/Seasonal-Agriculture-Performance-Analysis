```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')
df.head()
```

```python
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

df.isnull().sum()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Season', order=df['Season'].value_counts().index, palette='viridis')
plt.title('Number of Farms by Season')
plt.xlabel('Season')
plt.ylabel('Number of Farms')
plt.show()
```

Kharif has the highest number of farms in the dataset (1779), followed by Rabi (1627), while Zaid has the fewest at just 594. This makes sense since Zaid is a short summer season with limited farming activity, whereas Kharif and Rabi are the main cropping seasons in India.

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Season', y='Yield_Tonnes_Ha', estimator=np.mean, palette='viridis')
plt.title('Average Yield (Tonnes/Ha) by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.show()
```

Kharif season shows the highest average yield at around 5.6 tonnes/ha, followed by Rabi at approximately 5.05 tonnes/ha, while Zaid has the lowest average yield at around 4.6 tonnes/ha. The black vertical lines on each bar represent the confidence interval, and Zaid's wider interval reflects its smaller sample size (594 farms) compared to Kharif and Rabi.

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Season', y='Profit_INR', estimator=np.mean, palette='viridis')
plt.title('Average Profit (INR) by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.show()
```

Kharif shows by far the highest average profit at around ₹180,000, followed by Rabi at roughly ₹88,000. Zaid stands out here — its average profit is close to zero and even dips slightly negative, meaning many farms in the Zaid season are barely breaking even or running at a loss. This is a stronger seasonal difference than what we saw in yield, suggesting that even though Zaid's yield wasn't drastically lower, its economics (cost vs revenue) are much weaker compared to Kharif and Rabi.

```python
season_summary = df.groupby('Season')[['Total_Cost_INR', 'Revenue_INR', 'Profit_INR']].mean()
print(season_summary)
```

The breakdown shows that Zaid actually has the highest average cost (₹543,977) among all three seasons, even higher than Kharif and Rabi, while its average revenue is the lowest (₹519,172). This combination — higher cost with lower revenue — is what drives Zaid's average profit into negative territory (-₹24,805). In contrast, Kharif has both a reasonable cost and by far the highest revenue (₹710,719), which explains its strong profit margin. This suggests that farming in the Zaid season may involve higher input costs (possibly due to irrigation needs in the dry season) without a corresponding increase in output value.

```python
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Crop', y='Profit_INR', estimator=np.mean, palette='viridis')
plt.title('Average Profit (INR) by Crop')
plt.xlabel('Crop')
plt.ylabel('Average Profit (INR)')
plt.xticks(rotation=30)
plt.show()
```

There's a clear divide between low-value and high-value crops in this dataset. Sugarcane and Chilli stand out with the highest average profits (both above ₹700,000), while Wheat, Maize, and Rice actually show negative or near-zero average profit, meaning many farms growing these crops are operating at a loss. Pulses and Groundnut sit in a modest positive range, and Cotton shows a healthy mid-range profit. This suggests that crop choice has a much bigger impact on profitability than season alone, and that traditional staple crops like Wheat and Rice may need either cost optimization or better market pricing to become more profitable for farmers.

```python
plt.figure(figsize=(14,10))
numeric_cols = df.select_dtypes(include=[np.number]).drop(columns=['Farm_Area_Hectares'])
corr = numeric_cols.corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Heatmap of Agricultural Variables')
plt.show()
```

The heatmap reveals a few meaningful relationships beyond the obvious ones. Yield and Water Efficiency show a strong positive correlation (0.91), meaning farms that use water more efficiently tend to achieve significantly higher yields — this is a strong sustainability insight. Revenue and Profit are strongly correlated (0.89), which makes sense but confirms that revenue growth is the main profit driver rather than cost control. Interestingly, Rainfall shows a moderate positive correlation with Disease/Pest Risk (0.62), and Humidity also correlates with it (0.55), suggesting that wetter conditions increase pest and disease vulnerability. On the other hand, Pesticide usage has a negative correlation with Disease/Pest Risk (-0.47), which logically confirms that higher pesticide use helps control pest risk. Most environmental factors (temperature, soil pH, nutrients) show weak correlations with yield individually, indicating that yield is likely driven by a combination of factors rather than any single variable.

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Season', y='Profit_INR', palette='viridis')
plt.title('Profit Distribution by Season')
plt.xlabel('Season')
plt.ylabel('Profit (INR)')
plt.axhline(0, color='red', linestyle='--')
plt.show()
```

The boxplot reveals something the average profit chart didn't fully show — profit is highly variable within every season, with a large number of extreme outliers (some farms earning over ₹4,000,000 in profit, others losing over ₹1,000,000, regardless of season). Looking at the median (the line inside each box) confirms the earlier finding: Kharif's median sits clearly above the ₹0 line, Rabi's median is very close to ₹0, and Zaid's median is actually below the ₹0 line, meaning more than half of Zaid's farms are operating at a loss. This distribution view shows that while season influences the typical outcome, individual farm-level factors (crop choice, resource efficiency, cost management) still drive huge swings in profit within any single season.

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Season', y='Water_Efficiency_t_per_1000m3', estimator=np.mean, palette='viridis')
plt.title('Average Water Efficiency (Tonnes per 1000m3) by Season')
plt.xlabel('Season')
plt.ylabel('Water Efficiency (t/1000m3)')
plt.show()
```

Kharif shows the highest average water efficiency (around 5.9 t/1000m3), followed by Rabi (5.2), while Zaid has the lowest at roughly 4.4 t/1000m3. This pattern lines up directly with what we saw earlier — since Water Efficiency and Yield are strongly correlated (0.91 from the heatmap), Zaid's lower water efficiency likely contributes to its weaker yield and profit outcomes. This suggests Zaid season farming may involve less efficient irrigation practices, possibly because it falls in a hotter, drier period where more water is needed for the same output, and could be an area where better irrigation methods (like Drip irrigation) would especially help.

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Irrigation_Method', y='Profit_INR', estimator=np.mean, palette='viridis')
plt.title('Average Profit (INR) by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Average Profit (INR)')
plt.show()
```

Drip irrigation clearly stands out, giving an average profit of around ₹220,000 — nearly double or triple what the other methods produce. Flood irrigation gives the lowest average profit (₹73,000), while Rainfed (₹80,000) and Sprinkler (₹92,000) fall in between. This is a strong, actionable insight: since Drip irrigation is already known to use water more efficiently (less wastage compared to Flood irrigation), this result reinforces the earlier finding that water efficiency is strongly linked to yield and profit. It suggests that farmers still using Flood irrigation could significantly improve their profitability by switching to Drip systems.

```python
pivot = df.pivot_table(values='Profit_INR', index='Crop', columns='Season', aggfunc='mean')
plt.figure(figsize=(8,6))
sns.heatmap(pivot, annot=True, fmt='.0f', cmap='RdYlGn', center=0)
plt.title('Average Profit (INR): Crop vs Season')
plt.show()
```

This heatmap ties the whole analysis together and reveals a very clear pattern: for almost every crop, profit is highest in Kharif, drops in Rabi, and drops further in Zaid. Sugarcane grown in Kharif is the single most profitable combination in the entire dataset (average profit around 1,000,791), followed closely by Chilli in Kharif (954,380). On the other end, Rice, Wheat, and Maize are loss-making in every season, with losses getting progressively worse from Kharif to Zaid — Rice in Zaid, for example, averages a loss of about 227,239. This confirms that both crop choice and season matter together, not independently: high-value crops like Sugarcane and Chilli remain profitable across all seasons, while staple crops like Wheat, Rice, and Maize struggle regardless of season, though their losses are worst specifically in Zaid.

## Conclusion & Recommendations

Based on the seasonal analysis of this agricultural dataset, several clear patterns emerge. Kharif is consistently the strongest season across almost every metric — it has the highest average yield, the highest average profit, the highest revenue, and the best water efficiency. Rabi performs moderately well, while Zaid is the weakest season overall: despite having similar or only slightly lower yield, it suffers from higher input costs and lower revenue, resulting in an average loss. This shows that yield alone doesn't determine profitability — cost and revenue dynamics play an equally important role.

The correlation analysis revealed that water efficiency is one of the strongest drivers of yield (correlation of 0.91), and this was further confirmed by the finding that Drip irrigation produces significantly higher average profit than Flood, Rainfed, or Sprinkler methods. This points to irrigation efficiency as a key lever for improving farm profitability, especially in the Zaid season where water efficiency is lowest.

Crop choice turned out to be just as important as season. High-value crops like Sugarcane and Chilli remain profitable in every season, while staple crops like Wheat, Rice, and Maize are loss-making across all seasons, with losses worsening in Zaid. This suggests that season and crop selection interact — farmers growing low-margin staple crops in the Zaid season face the highest risk of financial loss.

Recommendations:

1.Farmers still using Flood irrigation should be encouraged to shift toward Drip irrigation, given its clear profitability advantage.

2.Zaid season farming needs closer support — either through better irrigation infrastructure or by encouraging higher-value crops (like Chilli) instead of loss-making staples during this season.

3.Wheat, Rice, and Maize cultivation may need cost optimization or improved market pricing support, as they show negative average profit across all seasons.

4.Further investigation could explore why Rainfall correlates with higher Disease/Pest Risk, to help farmers plan better pest management during wetter periods.

5.Since Cotton and Groundnut also showed positive average profit across seasons (as seen in the Crop vs Season heatmap), farmers currently growing consistently loss-making crops like Wheat or Rice could consider shifting toward these moderately profitable alternatives.

6.Since the correlation analysis showed pesticide usage has a negative correlation with disease/pest risk (-0.47), and rainfall/humidity increase that risk, farmers in high-rainfall seasons or regions should be advised to plan pesticide application more proactively to control the seasonal rise in disease/pest risk.

7.Since the boxplot showed huge profit variation within every season (not just between seasons), relying only on season-level averages can be misleading — recommend farm-level efficiency reviews (resource usage, cost management) rather than assuming all farms in a "good" season like Kharif are automatically profitable.

8.Since Water Efficiency showed the strongest correlation with Yield (0.91) among all factors tested, prioritizing irrigation efficiency improvements — not just fertilizer or seed quality — should be the primary focus area for boosting yield across all seasons, especially Zaid where water efficiency is lowest.
